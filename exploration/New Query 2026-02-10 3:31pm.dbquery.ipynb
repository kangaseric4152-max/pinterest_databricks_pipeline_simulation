```sql
with cte as (
    select 
      year(date_posted) as year_posted,
      count(year(date_posted)) as total_posts,
      
    from bright_data_pinterest_profiles_and_posts_datasets.datasets.pinterest_posts 
    where date_posted is not null 
    group by year(date_posted)
    order by year(date_posted)
)
select
  (select sort_array(array_agg(struct(year_posted, total_posts))) from cte) as total_posts_per_year, 
  count(*) as total_posts
from bright_data_pinterest_profiles_and_posts_datasets.datasets.pinterest_posts ;
-- join cte ON year(date_posted) = cte.year_posted
-- where date_posted is not null
-- group by year_posted
-- order by year_posted;
```
