```python
STAGING_PATH = "/Volumes/workspace/default/pinterest_pipeline/staging"
INBOX_PATH   = "/Volumes/workspace/default/pinterest_pipeline/inbox"

# List all year folders
years = dbutils.fs.ls(STAGING_PATH)

moved = False

for y in years:
    year_path = y.path
    
    quarters = dbutils.fs.ls(year_path)
    
    for q in quarters:
        source = q.path
        target = source.replace("/staging/", "/inbox/")
        
        # If target folder doesn't exist yet, move it
        try:
            dbutils.fs.ls(target)
            # Already exists → skip
        except:
            print(f"Moving batch: {source}")
            dbutils.fs.cp(source, target, recurse=True)
            moved = True
            break
    
    if moved:
        break

if not moved:
    print("No new batches to move.")
```
